# Escalado

Conjunto de datos: Iris (incluido en scikit-learn). Se comparan MinMaxScaler y StandardScaler.

```python
# Importar librerías necesarias
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Cargar el conjunto de datos
iris = load_iris(as_frame=True)
data = iris.data.copy()
data.head()
```

```python
# Estadísticas originales
print('Estadísticas originales:')
display(data.describe())
```

```python
# Aplicar MinMaxScaler
minmax_scaler = MinMaxScaler()
data_minmax = pd.DataFrame(minmax_scaler.fit_transform(data), columns=data.columns, index=data.index)

# Aplicar StandardScaler
standard_scaler = StandardScaler()
data_standard = pd.DataFrame(standard_scaler.fit_transform(data), columns=data.columns, index=data.index)
```

## Ejercicios

### 1. ¿Qué diferencias encuentras entre las estadísticas originales y las del escalado MinMax?

```python
# Comparar estadísticas antes y después de MinMaxScaler
print('Estadísticas originales:')
display(data.describe())
print('Estadísticas con MinMaxScaler:')
display(data_minmax.describe())
```

**Respuesta:** Los datos originales conservan sus unidades y rangos. MinMaxScaler transforma cada variable para que sus valores queden entre 0 y 1. Por eso, el mínimo es 0 y el máximo es 1; la media y la desviación también cambian.

### 2. Crea una gráfica de la distribución de `sepal length (cm)` antes y después del escalado.

```python
# Comparar la distribución de sepal length antes y después de MinMaxScaler
columna = 'sepal length (cm)'
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(data[columna], kde=True, ax=axes[0])
axes[0].set_title('Antes del escalado')
axes[0].set_xlabel('Longitud del sépalo (cm)')
sns.histplot(data_minmax[columna], kde=True, ax=axes[1], color='orange')
axes[1].set_title('Después de MinMaxScaler')
axes[1].set_xlabel('Longitud escalada')
plt.tight_layout()
plt.show()
```

### 3. Crea una gráfica de dispersión entre `sepal length (cm)` y `sepal width (cm)` con los datos escalados por StandardScaler.

```python
# Gráfica de dispersión con variables estandarizadas
plt.figure(figsize=(8, 5))
sns.scatterplot(x=data_standard['sepal length (cm)'], y=data_standard['sepal width (cm)'])
plt.title('Iris: sépalos después de StandardScaler')
plt.xlabel('Longitud del sépalo estandarizada')
plt.ylabel('Anchura del sépalo estandarizada')
plt.tight_layout()
plt.show()
```

### 4. ¿Qué técnica de escalado crees que es más adecuada para este conjunto de datos? ¿Por qué?

**Respuesta:** Ambas técnicas son adecuadas, dependiendo del modelo. Para Iris elegiría StandardScaler cuando se utilicen algoritmos sensibles a la escala, como KNN, SVM o regresión logística, porque centra cada variable en 0 y la deja con desviación estándar 1. MinMaxScaler es útil cuando se necesita limitar los valores al intervalo de 0 a 1. La elección final depende del algoritmo y del preprocesamiento requerido.
