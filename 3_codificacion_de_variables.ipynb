# 3. Codificación de variables

Conjunto de datos: Titanic (incluido en seaborn). Se convierten variables categóricas a valores numéricos.

```python
# Importar librerías necesarias
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
```

```python
# Cargar el conjunto de datos Titanic
data = sns.load_dataset('titanic')
data.head()
```

```python
# Seleccionar las columnas que se utilizarán
categorical_cols = ['sex', 'embarked', 'class']
data[categorical_cols].head()
```

## Ejercicios

### 1. ¿Qué diferencias encuentras entre LabelEncoder y OneHotEncoder?

**Respuesta:** LabelEncoder asigna un número entero a cada categoría (por ejemplo, 0 y 1). Es sencillo, pero en variables sin orden puede dar a entender que una categoría es mayor que otra. OneHotEncoder crea una columna binaria por categoría, evitando establecer un orden artificial. Se recomienda OneHotEncoder para variables nominales y LabelEncoder principalmente para etiquetas objetivo o categorías que realmente tengan un orden.

### 2. Crea una gráfica de barras comparando las frecuencias de `sex` antes y después de la codificación con LabelEncoder.

```python
# Codificar la variable sex con LabelEncoder
label_encoder = LabelEncoder()
data['sex_label'] = label_encoder.fit_transform(data['sex'])
print('Correspondencia de categorías:')
for categoria, numero in zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_)):
    print(f'{categoria}: {numero}')

# Comparar frecuencias antes y después de la codificación
frecuencias_originales = data['sex'].value_counts().sort_index()
frecuencias_codificadas = data['sex_label'].value_counts().sort_index()
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
frecuencias_originales.plot(kind='bar', ax=axes[0])
axes[0].set_title('Sexo original')
axes[0].set_xlabel('Categoría')
axes[0].set_ylabel('Frecuencia')
frecuencias_codificadas.plot(kind='bar', ax=axes[1], color='orange')
axes[1].set_title('Sexo codificado')
axes[1].set_xlabel('Etiqueta numérica')
axes[1].set_ylabel('Frecuencia')
plt.tight_layout()
plt.show()
```

### 3. Utiliza OneHotEncoder para codificar la columna `class`. ¿Qué ventajas tiene este enfoque frente a LabelEncoder?

```python
# Aplicar OneHotEncoder a la columna class
try:
    onehot_encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
except TypeError:  # Compatibilidad con versiones anteriores de scikit-learn
    onehot_encoder = OneHotEncoder(sparse=False, handle_unknown='ignore')

clases_codificadas = onehot_encoder.fit_transform(data[['class']])
columnas_onehot = onehot_encoder.get_feature_names_out(['class'])
data_onehot = pd.DataFrame(clases_codificadas, columns=columnas_onehot, index=data.index)
data = pd.concat([data, data_onehot], axis=1)
print('Columnas generadas:')
print(data_onehot.head())
```

**Ventajas:** OneHotEncoder representa cada clase con una columna independiente. Así, First, Second y Third no reciben valores que puedan interpretarse como distancias numéricas entre categorías. Es conveniente cuando la variable es categórica y no se desea imponer un orden.

### 4. Si quisieras aplicar un modelo de aprendizaje automático, ¿qué tipo de codificación elegirías para las variables categóricas? Explica tu respuesta.

**Respuesta:** Elegiría OneHotEncoder para variables nominales, como sexo o puerto de embarque, porque no tienen un orden natural. Para variables ordinales utilizaría una codificación que respete su orden. La codificación debe ajustarse al tipo de variable y al algoritmo para evitar relaciones numéricas incorrectas.
